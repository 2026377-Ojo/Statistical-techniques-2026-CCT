```python
df = pd.read_csv("Obesity.csv")

# Select only numerical columns
numeric_obesity = df.select_dtypes(include=['float64', 'int64'])

# Calculate the correlation matrix
correlation_matrix = numeric_obesity.corr()

# Set up the matplotlib figure
plt.figure(figsize=(12, 8))

# Create the heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    square=True,
    cbar_kws={"shrink": .8}
)

# Set the title
plt.title('Correlation Matrix for Obesity Dataset')

# Show the plot
plt.show()
```

```python
for col in ['Weight', 'Age']:
    population = df[col].dropna()

    for n in [5, 15, 25, 30, 35]:
        means = []
        for i in range(1000):
            sample = population.sample(n=n, replace=True)
            means.append(sample.mean())

        sns.histplot(means, kde=True, color='steelblue')
        plt.title(f'{col} Sampling Distribution: n = {n}')
        plt.xlabel(f'Sample Mean {col}')
        plt.show()
```

```python
std = population.std() # Larger samples give more stable sample means. Increasing n reduces the standard error at a square-root rate, so multiplying the sample size by four approximately halves the standard error.
 
for n in [5, 10, 20, 30, 50, 100]:
    se = std / np.sqrt(n)
    print(n, se)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Calculate the mean weight
mean_weight = df['Weight'].mean()

# Calculate the probability of weight being above the mean
p = (df['Weight'] > mean_weight).mean()

# Parameters
n = 10  # Number of randomly selected participants
k = np.arange(0, n + 1)  # Number of participants above the mean

# Binomial probability mass function
binom_pmf = binom.pmf(k, n, p)

# Plot the binomial distribution
plt.bar(k, binom_pmf, edgecolor='black')
plt.title('Binomial Distribution: Weight Above the Mean')
plt.xlabel('Number of Participants with Weight Above the Mean')
plt.ylabel('Probability')
plt.xticks(k)
plt.show()

# Probability that exactly 5 participants are above the mean
prob_5 = binom.pmf(5, n, p)
print(f'Mean weight: {mean_weight:.2f} kg')
print(f'Probability of weight above the mean: {p:.4f}')
print(f'Probability that exactly 5 participants are above the mean: {prob_5:.4f}')
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 10  # Number of participants
p = (df['family_history_with_overweight'] == 'yes').mean()
k = np.arange(0, n + 1)

# Binomial probability
binom_pmf = binom.pmf(k, n, p)

# Probability that exactly 3 participants have a family history
prob_3 = binom.pmf(3, n, p)

print(f'Probability of family history: {p:.4f}')
print(f'Probability that exactly 3 out of 10 have a family history: {prob_3:.4f}')
print(f'Percentage: {prob_3 * 100:.2f}%')

# Plot
plt.bar(k, binom_pmf, edgecolor='black')
plt.title('Binomial Distribution: Family History of Overweight')
plt.xlabel('Number of Participants with Family History')
plt.ylabel('Probability')
plt.xticks(k)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = (df['SMOKE'].astype(str).str.lower() == 'yes').mean()

k = np.arange(0, n + 1)
binom_pmf = binom.pmf(k, n, p)

prob_3 = binom.pmf(3, n, p)
print(df['SMOKE'].value_counts())

print(f'Probability of smoking: {p:.4f}')
print(f'Probability that exactly 3 out of 10 smoke: {prob_3:.4f}')
print(f'Percentage: {prob_3 * 100:.2f}%')

plt.bar(k, binom_pmf, edgecolor='black')
plt.title('Binomial Distribution: Smoking Status')
plt.xlabel('Number of Participants Who Smoke')
plt.ylabel('Probability')
plt.xticks(k)
plt.show()
```

```python
# Calculate lambda from the FAF(Physical activity Frequency) column
lambda_rate = df['FAF'].mean()
print(f"Lambda (average physical activity frequency): {lambda_rate:.2f}")

# Generate 1000 simulated values using the Poisson distribution
poisson_data_faf = np.random.poisson(lambda_rate, 1000)

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(
    poisson_data_faf,
    bins=np.arange(-0.5, poisson_data_faf.max() + 1.5, 1),
    kde=False,
    stat='probability'
)

plt.title('Poisson Distribution (Physical Activity Frequency)')
plt.xlabel('Number of Physical Activity Events')
plt.ylabel('Probability')
plt.grid()
plt.show()
```
