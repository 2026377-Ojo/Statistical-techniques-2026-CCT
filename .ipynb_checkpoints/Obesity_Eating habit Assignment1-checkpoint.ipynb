```python

Q1=df['Weight'].quantile(0.25)
Q3=df['Weight'].quantile(0.75)

IQR=Q3-Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_Weight=df[(df['Weight'] < lower_bound) | (df['Weight'] > upper_bound)]
print('lower bound: ', lower_bound)
print('upper bound: ', upper_bound)
print('number of outliers', len(outliers_Weight))
```

```python
print(outliers_Weight['Weight'])
```

```python
df = pd.read_csv("Obesity.csv")

# Select only numerical columns
numeric_obesity = df.select_dtypes(include=['float64', 'int64'])

# Calculate the correlation matrix
correlation_matrix = numeric_obesity.corr()

# Set up the matplotlib figure
plt.figure(figsize=(12, 8))

# Create the heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    square=True,
    cbar_kws={"shrink": .8}
)

# Set the title
plt.title('Correlation Matrix for Obesity Dataset')

# Show the plot
plt.show()
```

```python
for col in ['Weight', 'Age']:
    population = df[col].dropna()

    for n in [5, 15, 25, 30, 35]:
        means = []
        for i in range(1000):
            sample = population.sample(n=n, replace=True)
            means.append(sample.mean())

        sns.histplot(means, kde=True, color='steelblue')
        plt.title(f'{col} Sampling Distribution: n = {n}')
        plt.xlabel(f'Sample Mean {col}')
        plt.show()
```

```python
std = population.std() # Larger samples give more stable sample means. Increasing n reduces the standard error at a square-root rate, so multiplying the sample size by four approximately halves the standard error.
 
for n in [5, 10, 20, 30, 50, 100]:
    se = std / np.sqrt(n)
    print(n, se)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import binom

# Binomial Distribution
p = (df['family_history_with_overweight'] == 'yes').mean()

n = 10
k = 3

prob_binomial = binom.pmf(k, n, p)

print("Binomial Distribution")
print("Probability of exactly 3 people:", prob_binomial)
print("Percentage:", prob_binomial * 100, "%")
```

```python
print(df['family_history_with_overweight'].value_counts())
```

```python
from scipy.stats import binom
mean_weight = df['Weight'].mean()
p = (df['Weight'] >= mean_weight).mean()

# Binomial distribution
n = 10
k =5

prob = binom.pmf(k, n, p)


print("Mean Weight:", mean_weight)
print("Probability of success:", p)
print("Probability of exactly 5 people:", prob)
print("Percentage:", prob * 100, "%")
```

```python
from scipy.stats import binom
print(df['SMOKE'].value_counts())

p = (df['SMOKE'].astype(str).str.strip().str.lower().isin(['yes', 'true', '1']).mean())

n = 10
k = 3

prob = binom.pmf(k, n, p)

print("Binomial Distribution - SMOKE")
print("Probability of smoking:", p)
print("Sample size:", n)
print("Number of smokers:", k)
print("Probability of exactly 3 smokers:", prob)
print("Percentage:", prob * 100, "%")
```

```python
from scipy.stats import poisson


p = (df['NObeyesdad'] == 'Obesity_Type_I').mean()

lam = 10 * p

prob = poisson.pmf(k=2, mu=lam)

print("Expected count:", lam)
print("Probability:", prob)
print("Percentage:", prob * 100, "%")
```

```python
from scipy.stats import poisson


p = (df['NObeyesdad'] == 'Normal_Weight').mean()

lam = 10 * p

prob = poisson.pmf(k=1, mu=lam)

print("Expected count:", lam)
print("Probability:", prob)
print("Percentage:", prob * 100, "%")
```
